```python
def solution(A):
    if len(A) == 0:
        return 0

    A.sort()
    count = 1

    for i in range(1, len(A)):
        if A[i] != A[i - 1]:
            count += 1

    return count
```

```python
def solution(A):
    A.sort()
    n = len(A)

    product1 = A[n - 1] * A[n - 2] * A[n - 3]
    product2 = A[0] * A[1] * A[n - 1]

    return max(product1, product2)
```

```python
def solution(A):
    A.sort()

    left = 0
    right = len(A) - 1
    minimum = abs(A[0] + A[0])

    while left <= right:
        total = A[left] + A[right]

        if abs(total) < minimum:
            minimum = abs(total)

        if total < 0:
            left += 1
        elif total > 0:
            right -= 1
        else:
            return 0

    return minimum
```

```python
def solution(A):
    A.sort()
    n = len(A)
    count = 0

    for k in range(2, n):
        left = 0
        right = k - 1

        while left < right:
            if A[left] + A[right] > A[k]:
                count += right - left
                right -= 1
            else:
                left += 1

    return count
```

```python
def solution(K, M, A):
    low = max(A)
    high = sum(A)

    while low < high:
        mid = (low + high) // 2
        blocks = 1
        current_sum = 0

        for number in A:
            if current_sum + number > mid:
                blocks += 1
                current_sum = number
            else:
                current_sum += number

        if blocks > K:
            low = mid + 1
        else:
            high = mid

    return low
```
