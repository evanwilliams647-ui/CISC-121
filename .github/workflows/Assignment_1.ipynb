```python

def solution(A):
    # Convert the array into a set so we can
    # quickly check whether a number exists
    numbers = set(A)

    # Start with 1 because we need to find
    # the smallest missing positive integer
    missing_number = 1

    # Keep checking numbers until we find
    # one that is not in the set
    while missing_number in numbers:
        missing_number += 1

    # Return the first positive number
    # that was not found
    return missing_number
```

```python

def solution(A):
    # Keep track of how many eastbound cars
    # we have seen so far
    east_cars = 0

    # Store the total number of passing pairs
    passing_pairs = 0

    # Go through each car in the array
    for car in A:

        # A 0 represents a car going east
        if car == 0:
            east_cars += 1

        # A 1 represents a car going west
        else:
            # Every earlier eastbound car
            # forms a pair with this westbound car
            passing_pairs += east_cars

        # Codility requires -1 if the number
        # of passing pairs exceeds one billion
        if passing_pairs > 1000000000:
            return -1

    # Return the total number of passing pairs
    return passing_pairs
```

```python

def solution(A):
    # Calculate the sum of the entire array
    total_sum = sum(A)

    # Start the left side at zero
    left_sum = 0

    # Start with infinity so that the first
    # difference we calculate will be smaller
    min_difference = float('inf')

    # Loop through the array, excluding the last
    # element so that the right side is never empty
    for i in range(len(A) - 1):

        # Add the current number to the left side
        left_sum += A[i]

        # The right side is whatever remains
        # after subtracting the left side
        right_sum = total_sum - left_sum

        # Calculate the absolute difference
        # between the two sides
        difference = abs(left_sum - right_sum)

        # Update our minimum if we find
        # a smaller difference
        if difference < min_difference:
            min_difference = difference

    # Return the smallest difference found
    return min_difference
```

```python

def solution(A):
    # Create an empty dictionary to store
    # how many times each number appears
    counts = {}

    # Go through every number in the array
    for number in A:

        # If the number is already in the dictionary,
        # increase the number of occurrences
        if number in counts:
            counts[number] += 1

        # If it has not been seen yet,
        # add it to the dictionary with a count of 1
        else:
            counts[number] = 1

    # Check each number and its total count
    for number in counts:

        # The modulo operator (%) gives the remainder
        # when dividing by 2
        # An odd number will have a remainder of 1
        if counts[number] % 2 != 0:
            return number
```

```python

def solution(A, K):
    # If the array is empty, there is
    # nothing to rotate
    if len(A) == 0:
        return A

    # Use modulo to avoid doing extra rotations
    # For example, rotating an array of length 5
    # seven times is the same as rotating twice
    K = K % len(A)

    # Make a copy so we do not change
    # the original input array
    rotated = A.copy()

    # Repeat the rotation K times
    for i in range(K):

        # Remove the last element of the array
        last_element = rotated.pop()

        # Insert the removed element
        # at the beginning of the array
        rotated.insert(0, last_element)

    # Return the rotated array
    return rotated
```
